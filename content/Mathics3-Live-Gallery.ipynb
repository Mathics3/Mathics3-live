## Numeric Output

```python
(* Compute Euler's famous equation: *)
E ^ (Pi I)
```

```python
(* E is a Mathics3 symbolic constant whose value is precisely evaluated inside expressions. 
However, we can get its numeric approximation: *)
N[E]
```

```python
(* You can also explicitly specify the number of digits of accuracy.
Mathics3 supports arbitrarily large precision. *)
N[E, 30]
```

```python
(* Arbitrarily large integers are supported as well: *)
30! 
(* All Factorials over 5! have zeros at the end. *)
```

```python
(* The sum of the first 10 odd numbers is 11^2 *)
Sum[2 i +1, {i, 0, 10}]
```

```python
(* Powers of 2 *)
n = 8; 2 ^ # & /@ Range[0, n]
```
